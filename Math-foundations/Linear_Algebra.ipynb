# Linear Algebra Intuition

### Linear independence

How to find linear independence mathematically?
- A = [v1, v2, v3, ..]
- Row-reduce A (Apply Transformations)
- Find number of pivots ( leading number in non-zero row)
- if pivot = number of vectors, vectors are linearly independent
- if pivot < number of vectors, vectors are dependent

How to find linear independence in program?
```
    for each column:
        pivot = none
        find a row with a nonzero value
        swap it into pivot position
        for each next row below:
            factor = entry/pivot
            row = row - factor * pivot row
```

````python
import pandas as pd
import numpy as np
import math
````

````python
def is_linearly_independent(matrix):
    
    A = [] # creating a copy
    for row in matrix:
        new_row = row[:]
        A.append(new_row)
        
    cols = len(A[0])
    rows = len(A)
    rank=0
    
    for col in range(cols):
        pivot_row=None
        
        for row in range(rank,rows):
            if A[row][col] != 0:
                pivot_row = row
                break
        if pivot_row is None:
            continue

        """
        A = [[0,1,2]
             [1,2,3]] <- pivot row
        swap position to keep pivot row at the top, to avoid keeping tracking of pivot rows
            
        """
        # print(f"Swapping:{A[rank]} and {A[pivot_row]}")
        A[rank], A[pivot_row] = A[pivot_row], A[rank]
        
        # Eliminate entries below the pivot
        pivot = A[rank][col]
        for row in range(rank+1,rows):
            factor = A[row][col]/ pivot
            for j in range(col,cols):
                A[row][j] = A[row][j] - factor * A[rank][j]
        rank+=1
        print("Matrix after row reduction:",A)
    return rank == cols
A = [[1,2,3],
     [4,5,6],
     [7,8,9]]
B = [[0,1],
      [-1,0]]
is_linearly_independent(A)
````

### Dot product, projection and Gram Schmidt

````python
def dot(A,B):
    summation = 0
    cols = len(A)
    for col in range(cols):
        summation+= A[col]*B[col]
    return summation

def projection_of(A,B):
    cols = len(A)
    proj =[]
    for col in range(cols):
        proj.append(B[col])
    
    scalar = dot(A,B)/dot(B,B)
    for col in range(cols):
        proj[col]=scalar*proj[col]

    return proj

A = [1,0]
B = [0,1]
proj_A_on_B = projection_of(A,B)
print(proj_A_on_B)
    
````

````python
def normalize(A):
    magnitude = math.sqrt(dot(A,A))
    cols = len(A)
    new_vector  = []
    for col in range(cols):
        new_vector.append(A[col])

    for col in range(cols):
        new_vector[col] /= magnitude

    return new_vector

def gram_schmidt(matrix):
    orthonormal= []
    new_matrix = []
    rows = len(matrix)
    for row in matrix:
        new_matrix.append(row[:])

    for row in range(rows):
        v = new_matrix[row]
        cols = len(v)
        for u in orthonormal:
            project = projection_of(v,u)
            for col in range(cols):
                v[col]-=project[col]

        if dot(v,v) == 0:
            continue    
        u = normalize(v)
        orthonormal.append(u)
    return orthonormal
   
set_of_vectors = [
    [1,2,3],
    [1,1,0],
    [4,2,4]]
independency = is_linearly_independent(set_of_vectors)
print("Vectors are linearly independent?",independency)
if independency:
    orthonormal_vectors = gram_schmidt(set_of_vectors)
    print("Orthonormal Vectors : ",orthonormal_vectors)
````

### Shortcut version with Numpy

````python
# dot product 
a = np.array([1, 2, 3], dtype=float)
b = np.array([4, 5, 6], dtype=float)

print(np.dot(a, b))

# normalization
print(np.linalg.norm(a))

# projection of a onto b
proj = (np.dot(a, b) / np.dot(b, b)) * b
print(f"Projection of {a} onto {b}: {proj}")
````

````python
# Finding Rank and Linear Independency
A = [[1,2,3],[2,3,4]]
n = len(A)
rank = np.linalg.matrix_rank(A)
if rank == n:
    print("Vectors are linearly independent")
else :
    print("Vectors are not linearly independent")
````
