# Calculating Eigen values and Eigen Vectors

### Eigenvalues and Eigenvectors of a $2 \times 2$ Matrix

Consider a matrix

$$
A =
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
$$
An eigenvector $\mathbf{v}$ and its corresponding eigenvalue $\lambda$ satisfy

$$
A\mathbf{v} = \lambda\mathbf{v}
$$
#### 1. Finding the Eigenvalues
we move everything to one side:
$$
\det(A-\lambda I)=0
$$
For $$
A =
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
$$

we have

$$
A-\lambda I =
\begin{bmatrix}
a-\lambda & b \\
c & d-\lambda
\end{bmatrix}
$$
$$
(a-\lambda)(d-\lambda)-bc=0
$$
$$
ad-a\lambda-d\lambda+\lambda^2-bc=0
$$
$$
\lambda^2-(a+d)\lambda+(ad-bc)=0
$$

The trace of $A$ is $$\operatorname{tr}(A)=a+d $$ and the determinant is $$ \det(A)=ad-bc $$

Therefore, the characteristic equation can be written as

$$
\lambda^2-\operatorname{tr}(A)\lambda+\det(A)=0
$$
#### 2. Discriminant
The discriminant is$$ \Delta = \operatorname{tr}(A)^2-4\det(A)$$
The discriminant determines the type of eigenvalues:
$$
\Delta > 0
\quad\Rightarrow\quad
\text{two different real eigenvalues}
$$
$$
\Delta = 0
\quad\Rightarrow\quad
\text{repeated real eigenvalue}
$$
$$
\Delta < 0
\quad\Rightarrow\quad
\text{complex eigenvalues}
$$


#### 3. Eigenvalue Formula

Using the quadratic formula,

$$
\lambda_{1,2}
=
\frac{
\operatorname{tr}(A)
\pm
\sqrt{
\operatorname{tr}(A)^2-4\det(A)
}
}{2}
$$

#### 4. Finding the Eigenvectors

Once an eigenvalue $\lambda$ is known, its eigenvector is found by solving

$$
(A-\lambda I)\mathbf{v}=0
$$

where

$$
\mathbf{v}=
\begin{bmatrix}
x\\
y
\end{bmatrix}
$$

For a general $2\times2$ matrix,

$$
\begin{bmatrix}
a-\lambda&b\\
c&d-\lambda
\end{bmatrix}
\begin{bmatrix}
x\\
y
\end{bmatrix}
=
\begin{bmatrix}
0\\
0
\end{bmatrix}
$$

This produces the equations

$$
(a-\lambda)x+by=0
$$

and

$$
cx+(d-\lambda)y=0
$$
#### 5. Normalizing the Eigenvectors

The code normalizes each eigenvector so that its magnitude is $1$.


```python
import numpy as np
```

```python
def eigenvalue_2x2(A):
    a,b = A[0]
    c,d = A[1]
    trace = a + d
    determinant = a*d - b*c

    discriminant = np.pow(trace,2) - 4* determinant

    if discriminant < 0:
        print("Roots are imaginary")
    ## Need to add complex number code
        return 
    sqr_root_disc = np.sqrt(discriminant)

    # lambda^2 - trace(A) * lambda + det(A) = 0
    pos_root =(trace + sqr_root_disc) / 2
    neg_root =(trace - sqr_root_disc) / 2

    return [pos_root, neg_root]

def eigenvector_2x2(A, eigenvalue):
    a, b = A[0]
    c, d = A[1]

    '''
    [a-lambda     b  ] *[x] = 0
    [c       d-lambda]  [y]
    
    '''
    # (a - lambda)x + by = 0
    # Let's choose x = 1
    x = 1

    if b != 0:
        y = -(a - eigenvalue) * x / b
    elif c != 0:
        # If first equation doesn't help,
        # use cx + (d-lambda)y = 0
        y = 1
        x = -(d - eigenvalue) * y / c
    else:
        # b = 0 and c = 0
        # [a,0] [0,d] = [a-lambda , 0] [ 0, d-lambda]
        # det = a-lambda * d-lambda = 0
        # lambda = a or d, so A-lambda* I  = [0,0][0,d-a] or [a-d,0][0,0]
         if a == eigenvalue:
            # [0,0][0,d-a] * [x][y] = 0
            # d-a * y = 0
             # y = 0
            x = 1
            y = 0
         else:
            x = 0
            y = 1

     mag = np.sqrt((np.pow(x,2) + np.pow(y,2))) 
    return [x/mag, y/mag]
```

```python
A = np.array([[2,1],[2,3]])
eigen_values = eigenvalue_2x2(A)
print("Eigen Values:", eigen_values)
print()
for val in eigen_values:
    eigen_vector = eigenvector_2x2(A,val)
    AXv= A @ eigen_vector
    lambdaXv = []
    for ele in eigen_vector:
        lambdaXv.append(val*ele)

    print("A X v:",AXv)
    print("lambda X v:",lambdaXv)
    print("is A X v = lambda X v ?",np.allclose(AXv, lambdaXv))
    print()
```
